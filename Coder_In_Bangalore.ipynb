```python
with open("initialdata.txt", encoding="utf-8") as f:
    data = f.read()

chunks = data.split("\n\n")
chunks = [c for c in chunks if len(c) > 3]
```

```python
import pandas as pd

def convert_number(text):
    number = float(text.split()[0].replace(",", "").replace("K", "").replace("M", ""))

    if "K" in text:
        return int(number * 1000)
    elif "M" in text:
        return int(number * 1000000)
    return int(number)


def parse(chunk):
    lines = chunk.strip().split("\n")

    return {
        "Username": lines[0],
        "Posts": convert_number(lines[1]),
        "Followers": convert_number(lines[2]),
        "Following": convert_number(lines[3]),
        "Name": lines[4],
        "Type": lines[5],
        "Bio": "\n".join(lines[6:])
    }


# Parse all profiles
profiles = [parse(chunk) for chunk in chunks]

# Create DataFrame
df = pd.DataFrame(profiles)

# Save to CSV
df.to_csv("instagram_profiles.csv", index=False)

# Analysis
most_posted = max(profiles, key=lambda x: x["Posts"])
most_followed = max(profiles, key=lambda x: x["Followers"])
most_following = max(profiles, key=lambda x: x["Following"])

print("Most Posted:\n", most_posted)
print("\nMost Followed:\n", most_followed)
print("\nMost Following:\n", most_following)
```
